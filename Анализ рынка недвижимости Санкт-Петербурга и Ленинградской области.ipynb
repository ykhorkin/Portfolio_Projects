## Цель 
Агентству недвижимости из Петрозаводска, которое планирует выйти на рынок Санкт-Петербурга и Ленинградской области, нужен анализ объявлений о продаже жилой недвижимости в этих регионах, чтобы найти самые перспективные сегменты.

## Описание данных
База данных содержит 4 таблицы в схеме `real_estate`.

***Таблица*** `advertisement`
Содержит информацию об объявлениях:
1. **id** — идентификатор объявления (первичный ключ).
2. **first_day_exposition** — дата подачи объявления.
3. **days_exposition** — длительность нахождения объявления на сайте (в днях).
4. **last_price** — стоимость квартиры в объявлении, в руб.

***Таблица*** `flats`
Содержит информацию о квартирах:
1. **id** — идентификатор квартиры (первичный ключ, связан с первичным ключом `id` таблицы `advertisement`).
2. **city_id** — идентификатор города (внешний ключ, связан с `city_id` таблицы `city`).
3. **type_id** — идентификатор типа населённого пункта (внешний ключ, связан с `type_id` таблицы `type`).
4. **total_area** — общая площадь квартиры, в кв. метрах.
5. **rooms** — число комнат.
6. **ceiling_height** — высота потолка, в метрах.
7. **floors_total** — этажность дома, в котором находится квартира.
8. **living_area** — жилая площадь, в кв. метрах.
9. **floor** — этаж квартиры.
10. **is_apartment** — указатель, является ли квартира апартаментами (1 — является, 0 — не является).
11. **open_plan** — указатель, имеется ли в квартире открытая планировка (1 — открытая планировка квартиры, 0 — открытая планировка отсутствует).
12. **kitchen_area** — площадь кухни, в кв. метрах.
13. **balcony** — количество балконов в квартире.
14. **airports_nearest** — расстояние до ближайшего аэропорта, в метрах.
15. **parks_around3000** — число парков в радиусе трёх километров.
16. **ponds_around3000** — число водоёмов в радиусе трёх километров.

***Таблица*** `city`
Содержит информацию о городах:
1. **city_id** — идентификатор населённого пункта (первичный ключ).
2. **city** — название населённого пункта.

***Таблица*** `type`
Содержит информацию о типах населённых пунктов:
1. **type_id** — идентификатор типа населённого пункта (первичный ключ).
2. **type** — название типа населённого пункта.

**Запрос SQL**

```python
WITH limits AS (
                SELECT  
                      PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY total_area) AS total_area_limit,
                      PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY rooms) AS rooms_limit,
                      PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY balcony) AS balcony_limit,
                      PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY ceiling_height) AS ceiling_height_limit_h,
                      PERCENTILE_DISC(0.01) WITHIN GROUP (ORDER BY ceiling_height) AS ceiling_height_limit_l
                  FROM real_estate.flats     
               ),
filtered_id AS (
                SELECT id
                  FROM real_estate.flats  
                 WHERE 
                       total_area < (SELECT total_area_limit FROM limits)
                   AND (rooms < (SELECT rooms_limit FROM limits) OR rooms IS NULL)
                   AND (balcony < (SELECT balcony_limit FROM limits) OR balcony IS NULL)
                   AND ((ceiling_height < (SELECT ceiling_height_limit_h FROM limits)
                   AND ceiling_height > (SELECT ceiling_height_limit_l FROM limits)) OR ceiling_height IS NULL)
               ),
--выделение категории
   category AS ( 
                SELECT *,
                  CASE 
	                  WHEN f.city_id = '6X8I' THEN 'Санкт-Петербург'
	                  WHEN f.city_id <> '6X8I' THEN 'ЛенОбл'
	                END AS region,
	              CASE 
	   	              WHEN a.days_exposition BETWEEN 1 AND 30 THEN 'месяц'
	   	              WHEN a.days_exposition BETWEEN 31 AND 90 THEN 'квартал'
	   	              WHEN a.days_exposition BETWEEN 91 AND 180 THEN 'полгода'
	   	              WHEN a.days_exposition >= 181 THEN 'более полугода'
	   	              WHEN a.days_exposition IS NULL THEN 'еще не продано' --объявление еще активно
	                END AS activity,
	                       a.last_price / f.total_area AS price_m2
                  FROM real_estate.flats AS f
                  LEFT JOIN real_estate.advertisement AS a USING(id)
                 WHERE f.type_id = 'F8EM' AND id IN (SELECT * FROM filtered_id) --только города
               )
SELECT region,
       activity,
       COUNT(id) AS sum_advt,
       (COUNT(id)::REAL / SUM(COUNT(id)) OVER (PARTITION BY region))::NUMERIC(10, 2) AS part_advt,
       AVG(price_m2)::NUMERIC(10, 2)   AS avg_price_m2,
       AVG(total_area)::NUMERIC(10, 2) AS avg_total_area,
       PERCENTILE_DISC(0.5) WITHIN GROUP (ORDER BY rooms)   AS med_rooms,
       PERCENTILE_DISC(0.5) WITHIN GROUP (ORDER BY balcony) AS med_balcony,
       PERCENTILE_DISC(0.5) WITHIN GROUP (ORDER BY floor)   AS med_floor,
       AVG(kitchen_area)::NUMERIC(10, 2) AS avg_kitchen_area,
       AVG(is_apartment)::NUMERIC(10, 4) AS part_apartment,
       AVG(open_plan)::NUMERIC(10, 4) AS part_open_plan,
       AVG(parks_around3000)::NUMERIC(10, 4) AS avg_parks,
       AVG(ponds_around3000)::NUMERIC(10, 4) AS avg_pounds
  FROM category 
 GROUP BY region, activity
 ORDER BY sum_advt DESC 
; 
```

**Запрос SQL**

```python
WITH limits AS (
                SELECT  
                       PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY total_area) AS total_area_limit,
                       PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY rooms) AS rooms_limit,
                       PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY balcony) AS balcony_limit,
                       PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY ceiling_height) AS ceiling_height_limit_h,
                       PERCENTILE_DISC(0.01) WITHIN GROUP (ORDER BY ceiling_height) AS ceiling_height_limit_l
                  FROM real_estate.flats     
               ),
filtered_id AS (
                SELECT id
                  FROM real_estate.flats  
                 WHERE 
                       total_area < (SELECT total_area_limit FROM limits)
                   AND (rooms < (SELECT rooms_limit FROM limits) OR rooms IS NULL)
                   AND (balcony < (SELECT balcony_limit FROM limits) OR balcony IS NULL)
                   AND ((ceiling_height < (SELECT ceiling_height_limit_h FROM limits)
                   AND ceiling_height > (SELECT ceiling_height_limit_l FROM limits)) OR ceiling_height IS NULL)
               ),
date_activ  AS ( 
                SELECT *,
                       a.last_price / f.total_area AS price_m2,
                       (a.first_day_exposition + (a.days_exposition::TEXT || 'day')::INTERVAL)::DATE AS advt_removal_date
                  FROM real_estate.flats AS f
                  LEFT JOIN real_estate.advertisement AS a USING(id)
                 WHERE f.type_id = 'F8EM' -- только города
                   AND id IN (SELECT * FROM filtered_id)
                   AND first_day_exposition BETWEEN '2015-01-01' AND '2018-12-31' -- исключая неполные по данным года
                ), 
-- По дате публикации объявлений
 exposition AS  (
                 SELECT 
                        TO_CHAR(first_day_exposition, 'Month') AS month_exposition,
                        COUNT(id) AS sum_advt,
                        (COUNT(id)::real / (SELECT COUNT(id) FROM date_activ))::NUMERIC(10, 3) AS part_advt,
                        AVG(total_area)::NUMERIC(10, 2) AS avg_total_area,
                        AVG(price_m2)::NUMERIC(10, 2) AS avg_price_m2
                   FROM date_activ
                  GROUP BY month_exposition
                  ORDER BY sum_advt DESC
                ),
-- По дате снятия объявлений
    removal AS   ( 
                  SELECT 
                         TO_CHAR(advt_removal_date, 'Month') AS month_removal,
                         COUNT(id) AS sum_advt,
                         (COUNT(id)::real / (SELECT COUNT(id) FROM date_activ))::NUMERIC(10, 3) AS part_advt,
                         AVG(total_area)::NUMERIC(10, 2) AS avg_total_area,
                         AVG(price_m2)::NUMERIC(10, 2) AS avg_price_m2
                    FROM date_activ
                   WHERE advt_removal_date IS NOT NULL 
                   GROUP BY month_removal
                   ORDER BY sum_advt DESC 
                 )
 SELECT *
   FROM exposition AS e
   FULL JOIN removal AS r ON e.month_exposition = r.month_removal
;          
```

**Запрос SQL**

```python
WITH limits AS (
                SELECT  
                       PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY total_area) AS total_area_limit,
                       PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY rooms) AS rooms_limit,
                       PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY balcony) AS balcony_limit,
                       PERCENTILE_DISC(0.99) WITHIN GROUP (ORDER BY ceiling_height) AS ceiling_height_limit_h,
                       PERCENTILE_DISC(0.01) WITHIN GROUP (ORDER BY ceiling_height) AS ceiling_height_limit_l
                  FROM real_estate.flats     
               ),
filtered_id AS (
                SELECT id
                  FROM real_estate.flats  
                 WHERE 
                       total_area < (SELECT total_area_limit FROM limits)
                   AND (rooms < (SELECT rooms_limit FROM limits) OR rooms IS NULL)
                   AND (balcony < (SELECT balcony_limit FROM limits) OR balcony IS NULL)
                   AND ((ceiling_height < (SELECT ceiling_height_limit_h FROM limits)
                   AND ceiling_height > (SELECT ceiling_height_limit_l FROM limits)) OR ceiling_height IS NULL)
               ),
date_activ  AS ( 
                SELECT *,
                       a.last_price / f.total_area AS price_m2,
                       (a.first_day_exposition + (a.days_exposition::TEXT || 'day')::INTERVAL)::date AS advt_removal_date
                  FROM real_estate.flats AS f
                  LEFT JOIN real_estate.advertisement AS a USING(id)
                  LEFT JOIN real_estate.city AS c USING(city_id)
                  LEFT JOIN real_estate.type AS t USING(type_id)
                 WHERE f.city_id NOT IN ('6X8I') --без С-Петербурга
                   AND id IN (SELECT * FROM filtered_id)
                ) 
SELECT city,
  CASE 
     WHEN COUNT(id) BETWEEN 50 AND 100  THEN 'начальный'
     WHEN COUNT(id) BETWEEN 101 AND 200 THEN 'средний'
     WHEN COUNT(id) >= 201              THEN 'высокий'
   END AS activity_level,
       COUNT(id) AS sum_advt,
       AVG(total_area)::NUMERIC(10, 2) AS avg_total_area,
       AVG(price_m2)::NUMERIC(10, 2) AS avg_price_m2,
       (COUNT(advt_removal_date)::REAL / COUNT(id))::NUMERIC(10, 2) AS part_sales,
       AVG(days_exposition)::NUMERIC(10, 0) AS avg_duration_advt
  FROM date_activ
 GROUP BY city
HAVING COUNT(id) >= 50
 ORDER BY sum_advt DESC 
;
```
